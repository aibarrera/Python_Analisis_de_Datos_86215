## 1 · Variables
Una **variable** es un **nombre que apunta a un dato**. Es como una etiqueta (la cartela de una obra en un museo).
Para crearla se usa el signo `=`: a la izquierda, el nombre; a la derecha, el dato.

````python
ciudad = "Rosario"        # creamos la variable "ciudad" con un texto
anio = 2026               # creamos la variable "anio" con un número
print(ciudad)             # print() muestra el valor en pantalla
print(anio)
````

````python
# Una variable se puede CAMBIAR: la etiqueta pasa a apuntar a otro dato.
anio = 2026
print(anio)
anio = anio + 1           # toma el valor actual (2026), le suma 1 y lo guarda de nuevo
print(anio)               # 2027
anio += 1                 # forma abreviada de lo mismo: += suma y guarda
print(anio)               # 2028
````

## 2 · Tipos de datos
Cada dato tiene un **tipo**, que determina qué cosas se pueden hacer con él.

| Tipo | Nombre | Ejemplo | Se usa para |
|---|---|---|---|
| `int` | entero | `42`, `-3`, `1952` | contar, años, cantidades |
| `float` | decimal | `3.14`, `80.5` | medidas, precios, promedios |
| `str` | texto | `"hola"`, `'óleo'` | nombres, categorías |
| `bool` | booleano | `True`, `False` | sí / no |
| `NoneType` | "nada" | `None` | cuando **no hay dato** |

````python
# type() responde: "¿de qué tipo es este dato?"
print(type(42))           # int    : número sin punto
print(type(3.14))         # float  : número con punto decimal (el separador es el PUNTO, no la coma)
print(type("hola"))       # str    : está entre comillas
print(type(True))         # bool   : True o False, con la inicial en mayúscula y SIN comillas
print(type(None))         # NoneType
````

````python
# OJO: algo entre comillas es SIEMPRE texto, aunque parezca un número.
print(type(42))           # int
print(type("42"))         # str
print(10 + 5)             # 15      suma de números
print("10" + "5")         # "105"   con textos, + los PEGA (concatena)
````

### Convertir de un tipo a otro
| Función | Convierte a | Ejemplo | Resultado |
|---|---|---|---|
| `int()` | entero | `int("42")` | `42` |
| `float()` | decimal | `float("3.5")` | `3.5` |
| `str()` | texto | `str(42)` | `"42"` |
| `bool()` | booleano | `bool(0)` | `False` |

````python
print(int("42") + 1)          # 43   convertimos el texto en número y recién ahí sumamos
print(float("3.5") * 2)       # 7.0
print("Tengo " + str(30) + " años")   # sin str() daría error: no se pueden pegar texto y número
print(int(7.9))               # 7    int() CORTA los decimales (no redondea)
````

````python
# Si el valor no se puede convertir, aparece un error (ValueError). Lo mostramos sin cortar el notebook:
try:
    int("tres")
except ValueError as error:
    print("ValueError:", error)
````

````python
# bool(): cero, texto vacío, lista vacía y None valen False. Todo lo demás vale True.
print(bool(0), bool(""), bool([]), bool(None))      # False False False False
print(bool(5), bool("hola"), bool([0]))             # True True True
````

## 3 · Texto (`str`)

## 4 · Operadores
### Aritméticos
| Operador | Qué hace | Ejemplo | Resultado |
|---|---|---|---|
| `+` | suma | `7 + 3` | `10` |
| `-` | resta | `7 - 3` | `4` |
| `*` | multiplica | `7 * 3` | `21` |
| `/` | divide (da decimal) | `7 / 2` | `3.5` |
| `//` | división entera | `7 // 2` | `3` |
| `%` | resto de la división | `7 % 2` | `1` |
| `**` | potencia | `2 ** 3` | `8` |

````python
print(7 + 3, 7 - 3, 7 * 3)
print(7 / 2)              # 3.5  división normal
print(7 // 2)             # 3    división entera: solo la parte entera
print(7 % 2)              # 1    resto: 7 dividido 2 da 3 y SOBRA 1
print(2 ** 3)             # 8    2 elevado a la 3
print(2 + 3 * 4)          # 14   primero se multiplica, después se suma
print((2 + 3) * 4)        # 20   los paréntesis cambian el orden
````

### De comparación (el resultado siempre es `True` o `False`)
`==` igual · `!=` distinto · `>` mayor · `<` menor · `>=` mayor o igual · `<=` menor o igual

> ⚠️ **`=` asigna, `==` compara.** Son cosas distintas.

````python
print(5 == 5)             # True    ¿5 es igual a 5?
print(5 != 3)             # True    ¿5 es distinto de 3?
print(5 > 7)              # False
print("a" == "A")         # False   las mayúsculas cuentan
print("Py" in "Python")   # True    "in" pregunta si algo está dentro de otra cosa
````

### Lógicos: `and`, `or`, `not`
- `A and B` → `True` solo si **las dos** son verdaderas.
- `A or B` → `True` si **al menos una** es verdadera.
- `not A` → invierte el resultado.

````python
edad = 25
tiene_carnet = True
print(edad >= 18 and tiene_carnet)     # True   las dos se cumplen
print(edad < 18 or tiene_carnet)       # True   alcanza con una
print(not tiene_carnet)                # False  invierte
````

## 5 · Listas
Una **lista** guarda varios datos **en orden**, y se puede **modificar**. Se escribe con corchetes `[ ]`.

| Propiedad | Lista |
|---|---|
| ¿Tiene orden (índices)? | ✅ sí |
| ¿Se puede modificar? | ✅ sí |
| ¿Admite repetidos? | ✅ sí |

````python
obras = ["Paisaje", "Retrato", "Taller"]      # lista de tres textos
print(obras)
print(len(obras))          # 3    cantidad de elementos
print(obras[0])            # 'Paisaje'  PRIMER elemento: el índice empieza en 0
print(obras[2])            # 'Taller'
print(obras[-1])           # 'Taller'   el último (índice negativo)
print(obras[0:2])          # ['Paisaje', 'Retrato']   slicing: incluye el 0, NO incluye el 2
````

````python
obras = ["Paisaje", "Retrato", "Taller"]
obras.append("Mercado")         # append(x): agrega x AL FINAL
print(obras)
obras.insert(1, "Autorretrato") # insert(posición, x): agrega x en esa posición
print(obras)
````

````python
obras = ["Paisaje", "Retrato", "Taller", "Mercado"]
obras.remove("Taller")          # remove(x): quita el elemento que VALE x
print(obras)
ultimo = obras.pop()            # pop(): quita el ÚLTIMO elemento y lo devuelve
print(ultimo, obras)
obras[0] = "Paisaje del sur"    # modificar: se asigna a una posición
print(obras)
````

````python
numeros = [3, 1, 4, 1, 5]
print("Hay un 4:", 4 in numeros)       # in: ¿está en la lista?
print(numeros.count(1))                # 2    cuántas veces aparece el 1
print(numeros.index(4))                # 2    en qué posición está el 4
print(min(numeros), max(numeros), sum(numeros))    # 1 5 14   mínimo, máximo y suma
numeros.sort()                         # ordena la lista (la modifica)
print(numeros)                         # [1, 1, 3, 4, 5]
````

````python
# Error común: pedir una posición que no existe.
try:
    obras[10]
except IndexError as error:
    print("IndexError:", error)        # la lista no llega a esa posición
````

````python
# CUIDADO: b = a NO copia la lista, crea otra etiqueta para la MISMA lista.
a = [1, 2, 3]
b = a
b.append(4)
print(a)                  # [1, 2, 3, 4]   ¡también cambió a!
c = a.copy()              # .copy() sí crea una copia independiente
c.append(5)
print(a, c)
````

## 6 · Tuplas y diccionarios
### Tupla: como una lista que **no se puede modificar**
Se escribe con paréntesis `( )`.

````python
ubicacion = (2, 5)         # por ejemplo (sala, pared)
print(ubicacion[0])        # 2   se accede por índice, igual que en una lista
sala, pared = ubicacion    # desempaquetar: reparte los valores en variables
print(sala, pared)
try:
    ubicacion[0] = 9       # no se puede modificar
except TypeError as error:
    print("TypeError:", error)
````

### Diccionario: guarda pares **clave → valor**
Es como una **ficha**: cada dato tiene su nombre (clave). Se escribe con llaves `{ }`.
Se accede por **clave**, no por posición.

````python
ficha = {"titulo": "Taller", "anio": 1985, "tecnica": "fotografía"}
print(ficha["titulo"])               # 'Taller'   se pide por CLAVE
print(ficha.get("autor"))            # None       get(): si la clave no existe, no da error
print(ficha.get("autor", "sin dato"))# 'sin dato' y se puede dar un valor alternativo
print("anio" in ficha)               # True       in pregunta por las CLAVES
print(len(ficha))                    # 3          cantidad de pares
````

````python
ficha = {"titulo": "Taller", "anio": 1985}
ficha["expuesta"] = False            # clave NUEVA: se agrega
ficha["anio"] = 1986                 # clave EXISTENTE: se reemplaza el valor
print(ficha)
del ficha["expuesta"]                # del: elimina un par
print(ficha)
````

````python
print(ficha.keys())                  # las CLAVES
print(ficha.values())                # los VALORES
print(ficha.items())                 # los PARES (clave, valor)
for clave, valor in ficha.items():   # recorrer la ficha campo por campo
    print(clave, "->", valor)
````

````python
# Error común: pedir una clave que no existe.
try:
    ficha["autor"]
except KeyError as error:
    print("KeyError:", error)        # usá get() si no estás seguro de que exista
````

## 7 · Conjuntos (`set`)
Un **conjunto** guarda valores **sin repetir** y **sin orden**. Se escribe con llaves `{ }` con elementos.
Sirve para **quitar duplicados** y para comparar grupos.

````python
tecnicas = ["óleo", "grabado", "óleo", "acuarela", "grabado"]   # lista con repetidos
unicas = set(tecnicas)               # set() elimina los repetidos
print(unicas)                        # el orden puede cambiar: los conjuntos no tienen orden
print(len(unicas))                   # 3
unicas.add("fotografía")             # add(x): agrega un elemento
unicas.discard("grabado")            # discard(x): quita x si está (sin error si no está)
print("óleo" in unicas)              # True
````

````python
sala_1 = {"óleo", "acuarela"}
sala_2 = {"acuarela", "grabado"}
print(sala_1 & sala_2)               # en las DOS salas (intersección)
print(sala_1 | sala_2)               # en alguna de las dos, sin repetir (unión)
print(sala_1 - sala_2)               # en sala_1 y NO en sala_2 (diferencia)
````

> ⚠️ `{}` vacío crea un **diccionario**, no un conjunto. Para un conjunto vacío se escribe `set()`.

## 8 · Decidir y repetir: `if` y `for`
### `if` / `elif` / `else`: tomar decisiones
Los **dos puntos** `:` abren el bloque y la **sangría** (4 espacios) marca qué línea pertenece a él.

````python
edad = 17
if edad >= 18:                       # si se cumple esta condición...
    print("Mayor de edad")           # ...se ejecuta este bloque
elif edad >= 13:                     # si no, y se cumple esta otra...
    print("Adolescente")
else:                                # si no se cumplió ninguna de las anteriores
    print("Menor de 13")
````

### `for`: repetir para cada elemento

````python
obras = ["Paisaje", "Retrato", "Taller"]
for obra in obras:                   # en cada vuelta, "obra" toma un elemento distinto de la lista
    print(obra)                      # esta línea se ejecuta una vez por elemento
````

````python
# range(n) genera los números 0, 1, ..., n-1.   range(1, 4) genera 1, 2, 3 (el final NO se incluye).
for numero in range(1, 4):
    print(numero)
````

````python
# Patrón muy usado: ACUMULAR. Una variable que empieza en 0 y va sumando.
ventas = [120, 95, 130]
total = 0
for venta in ventas:
    total += venta                   # total = total + venta
print(total)                         # 345
````

````python
# for + if: contar cuántos elementos cumplen una condición.
edades = [15, 22, 17, 30, 41]
mayores = 0
for edad in edades:
    if edad >= 18:
        mayores += 1
print(mayores)                       # 3
````

## 🧭 Resumen: ¿qué colección uso?
| Colección | Se escribe | Orden | Se modifica | Repetidos | Usala para… |
|---|---|---|---|---|---|
| **Lista** | `[1, 2, 3]` | ✅ | ✅ | ✅ | una secuencia de elementos |
| **Tupla** | `(1, 2)` | ✅ | ❌ | ✅ | datos que no deben cambiar |
| **Diccionario** | `{"clave": valor}` | por clave | ✅ | claves únicas | una **ficha** con campos |
| **Conjunto** | `{1, 2}` | ❌ | ✅ | ❌ | valores únicos / comparar grupos |

### Machete de métodos
| Colección | Método | Qué hace |
|---|---|---|
| Lista | `append(x)` | agrega al final |
| Lista | `insert(i, x)` | agrega en la posición `i` |
| Lista | `remove(x)` | quita el valor `x` |
| Lista | `pop()` | quita el último y lo devuelve |
| Lista | `sort()` | ordena |
| Lista | `count(x)` / `index(x)` | cuántas veces aparece / en qué posición está |
| Diccionario | `get(k)` | valor de la clave (sin error si no existe) |
| Diccionario | `keys()` / `values()` / `items()` | claves / valores / pares |
| Diccionario | `d[k] = v` | agrega o reemplaza |
| Conjunto | `add(x)` / `discard(x)` | agrega / quita |
| Conjunto | `&` `\|` `-` | intersección / unión / diferencia |
| Texto | `upper()` `lower()` `strip()` `replace(a, b)` `split(s)` | transformaciones de texto |
| Todas | `len(x)` | cantidad de elementos |

Un error no es un fracaso: es Python explicándote qué pasó. **Leé la última línea**.

| Error | Qué significa | Qué revisar |
|---|---|---|
| `NameError` | usaste un nombre que no existe | ¿lo escribiste igual? ¿ejecutaste la celda que lo crea? |
| `TypeError` | mezclaste tipos o intentaste modificar algo que no se puede | ¿estás sumando texto con número? ¿modificás un texto o una tupla? |
| `ValueError` | el tipo está bien pero el valor no sirve | `int("hola")` |
| `IndexError` | pediste una posición que no existe | las listas empiezan en 0 |
| `KeyError` | pediste una clave que no existe en un diccionario | mayúsculas, tildes; o usá `get()` |
| `SyntaxError` | falta algo al escribir | un `:`, una comilla o un paréntesis |
| `IndentationError` | la sangría está mal | 4 espacios dentro del `if` o del `for` |
| `AssertionError` | en la autoevaluación, tu respuesta no coincide | leé el mensaje que acompaña |

## 📚 Glosario
| Palabra | Significa |
|---|---|
| **Variable** | nombre (etiqueta) que apunta a un dato |
| **Valor / dato** | la información que se guarda |
| **Tipo** | clase de dato: `int`, `float`, `str`, `bool` |
| **Función** | acción que recibe datos y devuelve algo: `len()`, `print()` |
| **Método** | función propia de un tipo de dato: `lista.append()` |
| **Argumento** | dato que se le pasa a una función o método |
| **Colección** | dato que agrupa otros: lista, tupla, diccionario, conjunto |
| **Índice** | posición de un elemento (empieza en 0) |
| **Clave** | nombre con el que se busca un valor en un diccionario |
| **Mutable** | que se puede modificar (lista, diccionario, conjunto) |
| **Inmutable** | que no se puede modificar (texto, número, tupla) |
| **Bucle** | repetición (`for`, `while`) |
| **Condición** | pregunta que da `True` o `False` |
| **Bloque** | líneas con sangría que pertenecen a un `if` o a un `for` |
