# 2 · Funciones: argumentos, `*args` y `**kwargs`

```python
def precio_final(precio, descuento=0, iva=0.21):
    return precio * (1 - descuento) * (1 + iva)

print(precio_final(18500))
print(precio_final(18500, descuento=0.10))
```

```python
def promedio(*numeros):
    return sum(numeros) / len(numeros)

print(promedio(10, 20, 30))
print(promedio(5, 15, 25, 35, 45))

def crear_ficha(nombre, **atributos):
    ficha = {"nombre": nombre}
    ficha.update(atributos)
    return ficha

print(crear_ficha("Remera", talle="M", color="rojo"))
```

# 3 · `map` y `filter`

```python
totales = list(map(total_venta, ventas))
print(totales[:5])

def es_de_palermo(venta):
    return venta["sucursal"] == "Palermo"

ventas_palermo = list(filter(es_de_palermo, ventas))
print(len(ventas_palermo))

totales_palermo = list(map(total_venta, filter(es_de_palermo, ventas)))
print(totales_palermo, sum(totales_palermo))
```

# 4 · `lambda`, `sorted` y `key=`

```python
print((lambda x: x * 2)(21))

totales = list(map(lambda v: v["cantidad"] * v["precio_unit"], ventas))
print(totales[:3])

top3 = sorted(ventas, key=lambda v: v["cantidad"] * v["precio_unit"], reverse=True)[:3]
for v in top3:
    print(v["id"], v["cantidad"] * v["precio_unit"])
```

```python
ventas_por_sucursal = {}
for v in ventas:
    ventas_por_sucursal[v["sucursal"]] = ventas_por_sucursal.get(v["sucursal"], 0) + total_venta(v)
print(ventas_por_sucursal)

ranking = sorted(ventas_por_sucursal.items(), key=lambda par: par[1], reverse=True)
print(ranking)

mayor = max(ventas, key=total_venta)
print(mayor["id"], total_venta(mayor))
```

# Glosario

| Término | En una línea |
|---|---|
| **Comprensión** | Expresión que construye una lista (`[]`), conjunto o diccionario (`{}`) en una línea |
| **Función** | Bloque de código con nombre, reutilizable |
| **Parámetro / argumento** | Variable en la definición / valor que se pasa en la llamada |
| **Argumento posicional / keyword** | Se asigna por orden / por nombre (`nombre=valor`) |
| **`*args` / `**kwargs`** | Capturan posicionales sobrantes en una tupla / keyword sobrantes en un diccionario |
| **Orden superior** | Función que recibe o devuelve otra función |
| **`map` / `filter`** | Aplica una función a cada elemento / se queda con los que dan `True` |
| **`lambda`** | Función anónima de una sola expresión |
| **`sorted` / `.sort()`** | Ordena devolviendo lista nueva / ordena la original (devuelve `None`) |
| **`key=`** | Función que define el criterio de orden (o de máximo/mínimo) |
| **Módulo / paquete** | Archivo `.py` reutilizable / carpeta de módulos |
| **`import`, `as`, `from`** | Traer un módulo, darle alias, traer solo una parte |
| **`pip` / `conda`** | Gestores de paquetes |
