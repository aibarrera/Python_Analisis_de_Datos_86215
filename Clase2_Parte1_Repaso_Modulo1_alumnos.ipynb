## 3.5 · Referencias y copias

📘 **Objetos** (objetos y referencia) y **Operaciones** (identidad: `is`).

`b = a` **no copia**: crea otro nombre para el **mismo objeto**. Si es mutable (lista, diccionario, conjunto), un cambio hecho desde un nombre se ve desde el otro. Es como compartir el link de una planilla en vez de hacer una copia. Para copiar: `.copy()`.

````python
meta_a = [100, 200, 300]
meta_b = meta_a
meta_b.append(400)
print(meta_a)
print(meta_a is meta_b, id(meta_a) == id(meta_b))
````

````python
original = [100, 200, 300]
copia = original.copy()
copia.append(400)
print(original)
print(copia)
print(original == copia[:3], original is copia)
````

## 4.1 · Condicional: `if` / `elif` / `else`

## 4.4 · Listas por comprensión

📘 **Módulo 1 · Estructuras de control** (Listas por comprensión).

Arman una colección nueva en **una sola expresión**:

```
[ expresión   for elemento in iterable   if condición ]
```

- La **expresión** transforma cada elemento; el `if` (opcional) **filtra**.
- Si querés elegir entre **dos valores**, el `if/else` va **al principio**: `[a if cond else b for x in it]`.
- Con `{}` arman conjuntos y, con `clave: valor`, diccionarios.
- Es un `for` + `append` compactado: si no se lee de un vistazo, usá el `for`.

````python
totales_for = []
for v in ventas:
    totales_for.append(v["cantidad"] * v["precio_unit"])

totales = [v["cantidad"] * v["precio_unit"] for v in ventas]

print(totales_for == totales)
print(totales)
````

````python
ids_belgrano = [v["id"] for v in ventas if v["sucursal"] == "Belgrano"]
print(ids_belgrano)

etiquetas = ["alta" if t >= 100000 else "normal" for t in totales]
print(etiquetas)

productos_distintos = {v["producto"] for v in ventas}
print(productos_distintos)

total_por_id = {v["id"]: v["cantidad"] * v["precio_unit"] for v in ventas}
print(total_por_id[1006])
````

## 4.5 · Diccionario de frecuencias (acumulador por grupo)

📘 **Estructuras de control** (Ejemplo: diccionario de frecuencias).

Muchas preguntas de negocio son del tipo *"¿cuánto … por …?"*. Se resuelven con un diccionario que va acumulando: si la clave **existe**, se suma; si **no**, se empieza. `dict.get(clave, 0)` simplifica ese paso. Es lo que más adelante hace `groupby` en pandas.

````python
ventas_por_sucursal = {}
for v in ventas:
    monto = v["cantidad"] * v["precio_unit"]
    sucursal = v["sucursal"]
    if sucursal in ventas_por_sucursal:
        ventas_por_sucursal[sucursal] += monto
    else:
        ventas_por_sucursal[sucursal] = monto
print(ventas_por_sucursal)
````

````python
unidades_por_producto = {}
for v in ventas:
    producto = v["producto"]
    unidades_por_producto[producto] = unidades_por_producto.get(producto, 0) + v["cantidad"]
print(unidades_por_producto)
````

````python
# La misma idea con una comprensión (más compacta, pero recorre la lista una vez por producto)
productos_vendidos = [v["producto"] for v in ventas]
frecuencia = {p: productos_vendidos.count(p) for p in set(productos_vendidos)}
print(frecuencia)
````

# Glosario

| Término | En una línea |
|---|---|
| **Dato** | Representación simbólica de un aspecto de la realidad |
| **Tipo** | Qué valores admite un dato y qué se puede hacer con él (`int`, `float`, `str`, `bool`…) |
| **Variable** | Nombre que apunta a un objeto en memoria |
| **Tipado dinámico** | El tipo lo tiene el valor, no la variable |
| **Case sensitive** | Distingue mayúsculas de minúsculas |
| **snake_case** | Convención: minúsculas separadas por `_` |
| **Objeto** | Todo lo que existe en Python: números, textos, listas, funciones… |
| **Mutable / inmutable** | Se puede / no se puede modificar sin crear otro objeto |
| **Iterable** | Algo que se puede recorrer con `for` |
| **Índice** | Posición de un elemento (empieza en 0) |
