# 1 · ¿Por qué NumPy? Rendimiento y memoria

````python
np.random.seed(0)
n = 1_000_000
precios_m   = np.random.randint(5000, 150000, n)
unidades_m  = np.random.randint(1, 6, n)

print(precios_m.size)
print(precios_m.dtype)
````

````python
import time

lista_precios  = precios_m.tolist()
lista_unidades = unidades_m.tolist()

inicio = time.perf_counter()
totales_lista = []
for p, u in zip(lista_precios, lista_unidades):
    totales_lista.append(p * u * 1.21)
t_for = time.perf_counter() - inicio

inicio = time.perf_counter()
totales_np = precios_m * unidades_m * 1.21
t_np = time.perf_counter() - inicio

print(f"for:   {t_for:.4f} s")
print(f"NumPy: {t_np:.4f} s")
print(f"NumPy fue ~{t_for / t_np:.0f} veces más rápido")

print(np.allclose(totales_lista, totales_np))
````

````python
import sys

bytes_punteros = sys.getsizeof(lista_precios)
bytes_objetos = sum(sys.getsizeof(x) for x in lista_precios)
print(f"Lista de Python: {(bytes_punteros + bytes_objetos) / 1_000_000:.1f} MB")

print(f"Array int64:     {precios_m.nbytes / 1_000_000:.1f} MB")

precios_32 = precios_m.astype(np.int32)
print(f"Array int32:     {precios_32.nbytes / 1_000_000:.1f} MB")

unidades_8 = unidades_m.astype(np.int8)
print(f"Array int8:      {unidades_8.nbytes / 1_000_000:.1f} MB")
````

# Glosario

| Término | En una línea |
|---|---|
| **NumPy** | Librería de cálculo numérico sobre arrays; base de pandas y Matplotlib |
| **Array (`ndarray`)** | Colección multidimensional **homogénea** y de **tamaño fijo**, en memoria contigua |
| **`dtype`** | Tipo de dato de los elementos del array |
| **`shape` / `ndim` / `size`** | Forma (tupla) / cantidad de dimensiones / cantidad de elementos |
| **`nbytes` / `strides`** | Bytes que ocupan los datos / bytes para avanzar una posición por dimensión |
| **`reshape`** | Cambia la forma sin cambiar los datos (el total debe coincidir) |
| **Indexing / slicing** | Seleccionar con corchetes: `a[f, c]`, `a[:, 2]`, `a[1:3]` |
| **Vista / copia** | Comparte datos con el original / tiene datos propios |
| **Vectorización** | Operar sobre todos los elementos sin bucles explícitos |
| **Función universal (ufunc)** | Función que opera elemento a elemento (`np.sqrt`, `np.log`…) |
| **Broadcasting** | Operar entre arrays de distinta forma "estirando" las dimensiones iguales a 1 |
| **Máscara booleana** | Array de `True`/`False` usado para filtrar: `a[a > 5]` |
| **`&`, `\|`, `~`** | "y", "o", "no" entre arrays de booleanos (cada condición entre paréntesis) |
| **Agregación** | Función que reduce un conjunto a un solo valor (`sum`, `mean`, `std`…) |
| **`axis`** | Dimensión sobre la que se agrega (`0` = por columna, `1` = por fila en una matriz) |
| **`NaN`** | Valor faltante; `nanmean`, `nansum`… lo ignoran |
| **Producto escalar** | Suma de los productos elemento a elemento: `v @ w` |
| **Producto matricial** | Fila × columna con `@`; columnas de la 1.ª = filas de la 2.ª |
| **`np.linalg.solve`** | Resuelve sistemas de ecuaciones lineales `A x = b` |
