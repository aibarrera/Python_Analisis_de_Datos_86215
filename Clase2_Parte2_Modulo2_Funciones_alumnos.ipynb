# 1 · Funciones

📘 **Módulo 2 · Funciones** (Definición).

`def nombre(parámetros):` y el cuerpo con sangría. `return` indica el valor de retorno; si no hay `return`, la función devuelve `None`.

```python
def calcular_iva(precio):
    """Devuelve el IVA (21%) de un precio neto."""
    return precio * 0.21

print(calcular_iva(18500))

iva_remera = calcular_iva(18500)
print(iva_remera)
print(iva_remera + 18500)

for producto, precio in [("Remera", 18500), ("Jean", 42000), ("Mochila", 35000)]:
    print(producto, "->", calcular_iva(precio))
```

### `return` vs `print`

`return` **entrega** un valor para seguir usándolo; `print` solo lo **muestra** en pantalla y la función devuelve `None`.

```python
def iva_con_print(precio):
    print(precio * 0.21)

def iva_con_return(precio):
    return precio * 0.21

a = iva_con_print(1000)
b = iva_con_return(1000)
print("a =", a)
print("b =", b)
```

```python
# ⚠️ Esta celda da error a propósito
a + 1
```

```python
def probar_return():
    return "salí"
    print("esta línea nunca se ejecuta: return termina la función")

print(probar_return())
```

### Decisiones y varios valores de retorno

Una función puede usar `if`, bucles y todo lo visto. Con `return a, b, c` devuelve una **tupla**.

```python
def clasificar_ticket(monto):
    """Clasifica un ticket según su monto: 'bajo', 'medio' o 'alto'."""
    if monto < 20000:
        return "bajo"
    elif monto < 80000:
        return "medio"
    return "alto"

for monto in [5000, 62000, 150000]:
    print(f"${monto:>7,} -> {clasificar_ticket(monto)}")
```

```python
def resumen_precios(precios):
    """Devuelve (mínimo, máximo, promedio) de una lista de precios."""
    return min(precios), max(precios), sum(precios) / len(precios)

resultado = resumen_precios([18500, 42000, 35000, 120000])
print(resultado, type(resultado))

minimo, maximo, promedio = resumen_precios([18500, 42000, 35000, 120000])
print(f"Mín: {minimo:,} | Máx: {maximo:,} | Promedio: {promedio:,.0f}")
```

### Posicionales, keyword y valores por defecto

- **Posicionales:** se asignan por orden y son obligatorios.
- **Keyword:** `nombre=valor`, en cualquier orden. Los que tienen **valor por defecto** son opcionales.
- Los posicionales van **antes** que los keyword.

```python
def precio_final(precio, descuento=0, iva=0.21):
    """Precio con descuento e IVA. 'descuento' e 'iva' son opcionales."""
    return precio * (1 - descuento) * (1 + iva)

print(precio_final(1000))
print(precio_final(1000, 0.10))
print(precio_final(1000, iva=0.105))
print(precio_final(iva=0.105, precio=1000, descuento=0.2))
```

### `*args` y `**kwargs`

- `*args` captura **todos los posicionales sobrantes** en una **tupla**.
- `**kwargs` captura **todos los keyword sobrantes** en un **diccionario** (`nombre → valor`) y va **al final**.

El nombre es convención; lo importante son los asteriscos.

```python
def promedio(*numeros):
    """Promedio de una cantidad cualquiera de números."""
    print("Recibí:", numeros, type(numeros))
    if not numeros:
        return None
    return sum(numeros) / len(numeros)

print(promedio(10, 20, 30))
print(promedio(5, 8, 9, 10, 4, 7, 6))
print(promedio())
```

```python
def crear_ficha(nombre, **atributos):
    """Arma la ficha de un producto con los atributos que se pasen."""
    print("Atributos recibidos:", atributos, type(atributos))
    ficha = {"nombre": nombre}
    ficha.update(atributos)
    return ficha

print(crear_ficha("Remera", color="azul", talle="M"))
print(crear_ficha("Mochila", capacidad_litros=25, impermeable=True, marca="Andes"))
```

### 💬 ¿Qué imprime?

```python
def mostrar(a, b=2, *args, **kwargs):
    print("a =", a, "| b =", b, "| args =", args, "| kwargs =", kwargs)

mostrar(1)
mostrar(1, 5)
mostrar(1, 5, 9, 10)
mostrar(1, 5, 9, 10, color="rojo")
```

### Los argumentos se pasan por referencia

📘 **Módulo 2 · Funciones** (Argumentos). Si pasás un objeto **mutable** y la función lo modifica, el cambio se ve **afuera**. Mejor devolver una lista nueva que modificar la original.

```python
def aplicar_descuento_mal(precios, descuento):
    """Modifica la lista ORIGINAL (efecto secundario)."""
    for i in range(len(precios)):
        precios[i] = precios[i] * (1 - descuento)

def aplicar_descuento_bien(precios, descuento):
    """Devuelve una lista NUEVA y deja la original intacta."""
    return [p * (1 - descuento) for p in precios]

lista_precios = [100, 200, 300]

nueva = aplicar_descuento_bien(lista_precios, 0.10)
print("original:", lista_precios, "| nueva:", nueva)

aplicar_descuento_mal(lista_precios, 0.10)
print("original después de la 'mala':", lista_precios)
```

| | `lista.sort()` | `sorted(iterable)` |
|---|---|---|

# 4 · Funciones de orden superior

📘 **Módulo 2 · Funciones de orden superior** (`map` y `filter`).

Son funciones que **reciben otra función** como parámetro o **devuelven** una. En Python las funciones también son objetos: se pueden asignar a variables y pasar como argumento.

```python
def calcular_iva(precio):
    return precio * 0.21

print(calcular_iva)
otra_referencia = calcular_iva
print(otra_referencia(1000))

def mitad(x):
    return x / 2

operaciones = {
    "iva": calcular_iva,
    "mitad": mitad,
}
print(operaciones["iva"](1000))
```

Una función de orden superior hecha a mano: aplica **cualquier función** a **cualquier lista**.

```python
def aplicar(funcion, coleccion):
    """Aplica 'funcion' a cada elemento y devuelve la lista de resultados."""
    resultado = []
    for elemento in coleccion:
        resultado.append(funcion(elemento))
    return resultado

precios = [18500, 42000, 35000]

print(aplicar(calcular_iva, precios))
print(aplicar(round, [1.4, 2.6, 3.5]))
print(aplicar(len, ["Palermo", "Belgrano"]))
```

### `map(función, iterable)`

Aplica la función **a cada elemento**. Devuelve un generador (se calcula cuando se lo pide), por eso lo envolvemos con `list()`.

```python
totales = list(map(total_venta, ventas))
print(totales)
```

```python
m = map(total_venta, ventas)
print(m)
print(list(m)[:3])
print(list(m))
```

### `filter(función, iterable)`

Se queda con los elementos para los que la función devuelve `True`.

```python
def es_de_palermo(venta):
    return venta["sucursal"] == "Palermo"

ventas_palermo = list(filter(es_de_palermo, ventas))
print(len(ventas_palermo), "ventas en Palermo")
print([v["id"] for v in ventas_palermo])
```

```python
totales_palermo = list(map(total_venta, filter(es_de_palermo, ventas)))
print(totales_palermo)
print(sum(totales_palermo))
```

### Comparación con las comprensiones

Casi todo lo que hace `map` y `filter` también se hace con una comprensión. El **concepto** (pasar una función para aplicarla a los datos) reaparece en `sorted(key=...)` y, más adelante, en `df.apply(función)` de pandas.

```python
print(list(map(total_venta, ventas)) == [total_venta(v) for v in ventas])
print(list(filter(es_de_palermo, ventas)) == [v for v in ventas if es_de_palermo(v)])
```

`max(iterable, key=función)` devuelve el elemento cuyo resultado de `key` sea el mayor: *¿cuál fue la venta más grande?* sin armar el bucle.

```python
venta_mayor = max(ventas, key=total_venta)
print(venta_mayor)
print(f"Total: ${total_venta(venta_mayor):,}")

venta_menor = min(ventas, key=total_venta)
print(venta_menor["id"], venta_menor["producto"], total_venta(venta_menor))
```

# 5 · Funciones anónimas (`lambda`) y `sorted`

📘 **Módulo 2 · Funciones anónimas** (lambda, `sorted`, método `sort`).

`lambda parámetros: expresión` es una función **sin nombre**, de **una sola expresión**, que se define "al paso" donde se necesita (típicamente dentro de `map`, `filter`, `sorted`). Si la lógica es más compleja, usá `def`.

```python
print((lambda x: x * 2)(21))

doble = lambda x: x * 2
print(doble(10))
print(type(doble))
```

```python
totales = list(map(lambda v: v["cantidad"] * v["precio_unit"], ventas))
print(totales[:5])

ventas_belgrano = list(filter(lambda v: v["sucursal"] == "Belgrano", ventas))
print([v["id"] for v in ventas_belgrano])

productos_caros = list(map(lambda v: v["producto"],
                           filter(lambda v: v["cantidad"] * v["precio_unit"] > 50000, ventas)))
print(productos_caros)
```

```python
etiquetas = list(map(lambda t: "alta" if t >= 100000 else "normal", totales))
print(etiquetas)
```

### `sorted(iterable, key=..., reverse=...)`

Devuelve una **lista nueva** ordenada. `key` es **una función** que define el criterio de orden y `reverse=True` ordena de mayor a menor. El método `.sort()` ordena la lista **original** y devuelve `None`.

```python
numeros = [9, 3, 7, 1]
print(sorted(numeros))
print(sorted(numeros, reverse=True))

nombres = ["Zoe", "ana", "Bruno", "carla"]
print(sorted(nombres))
print(sorted(nombres, key=str.lower))
print(sorted(nombres, key=len))
```

```python
top_3 = sorted(ventas, key=lambda v: v["cantidad"] * v["precio_unit"], reverse=True)[:3]
for v in top_3:
    print(v["id"], v["producto"], f"${total_venta(v):,}")
```

### Ordenar un diccionario

`.items()` da pares `(clave, valor)` y la lambda elige por cuál ordenar: *"armame un ranking de sucursales por ventas"*.

```python
ventas_por_sucursal = {}
for v in ventas:
    ventas_por_sucursal[v["sucursal"]] = ventas_por_sucursal.get(v["sucursal"], 0) + total_venta(v)

print(ventas_por_sucursal)
print(list(ventas_por_sucursal.items()))

ranking = sorted(ventas_por_sucursal.items(), key=lambda par: par[1], reverse=True)
for posicion, (sucursal, monto) in enumerate(ranking, start=1):
    print(f"{posicion}. {sucursal:<9} ${monto:>9,}")
```

```python
ventas_crudas = [
    {"id": 1, "sucursal": "  palermo",  "producto": "remera",   "cantidad": "2", "precio": "$ 18.500,00"},
    {"id": 2, "sucursal": "BELGRANO ",  "producto": "JEAN",     "cantidad": "1", "precio": "$42.000,00"},
    {"id": 3, "sucursal": "Centro",     "producto": "Mochila",  "cantidad": "3", "precio": "$ 35.000,50"},
    {"id": 4, "sucursal": "palermo",    "producto": " gorra ",  "cantidad": "5", "precio": "12000"},
    {"id": 5, "sucursal": " Belgrano",  "producto": "remera",   "cantidad": "4", "precio": "$ 18.500,00"},
]

def parsear_precio(texto):
    """'$ 18.500,50' -> 18500.5"""
    limpio = texto.replace("$", "").replace(".", "").replace(",", ".").strip()
    return float(limpio)

def limpiar_texto(texto):
    """Quita espacios y deja formato 'Capitalizado'."""
    return texto.strip().capitalize()

def limpiar_venta(venta_cruda):
    """Recibe una venta cruda y devuelve una NUEVA con los tipos y el formato correctos."""
    return {
        "id": venta_cruda["id"],
        "sucursal": limpiar_texto(venta_cruda["sucursal"]),
        "producto": limpiar_texto(venta_cruda["producto"]),
        "cantidad": int(venta_cruda["cantidad"]),
        "precio": parsear_precio(venta_cruda["precio"]),
    }

ventas_limpias = list(map(limpiar_venta, ventas_crudas))
for v in ventas_limpias:
    print(v)
```

### Una función reutilizable: "total por …"

Una sola función responde *"¿cuánto … por …?"* para **cualquier** columna: el nombre del campo es un parámetro (conceptualmente, lo que hace `groupby` en pandas).

```python
def total_por(filas, campo):
    """Suma el total (cantidad * precio_unit) agrupando por el campo indicado."""
    acumulado = {}
    for fila in filas:
        clave = fila[campo]
        acumulado[clave] = acumulado.get(clave, 0) + total_venta(fila)
    return acumulado

def ranking(diccionario, n=3):
    """Devuelve los n pares (clave, valor) con mayor valor."""
    return sorted(diccionario.items(), key=lambda par: par[1], reverse=True)[:n]

print("Por sucursal: ", total_por(ventas, "sucursal"))
print("Por categoría:", total_por(ventas, "categoria"))
print("Por producto: ", total_por(ventas, "producto"))
print()
print("Top 3 productos por facturación:", ranking(total_por(ventas, "producto")))
```

### Reporte final

```python
def reporte(filas):
    """Imprime un reporte resumido de las ventas."""
    totales = list(map(total_venta, filas))
    print("=" * 46)
    print(f"{'REPORTE MODASUR':^46}")
    print("=" * 46)
    print(f"Ventas registradas: {len(filas)}")
    print(f"Facturación total:  ${sum(totales):,}")
    print(f"Ticket promedio:    ${sum(totales) / len(totales):,.0f}")
    print(f"Clientes distintos: {len({f['cliente'] for f in filas})}")
    print("-" * 46)
    print("Ranking de sucursales:")
    for i, (sucursal, monto) in enumerate(ranking(total_por(filas, "sucursal"), n=3), start=1):
        print(f"  {i}. {sucursal:<10} ${monto:>10,}  ({monto / sum(totales):.1%})")
    print("-" * 46)
    mejor = max(filas, key=total_venta)
    print(f"Mayor venta: id {mejor['id']} ({mejor['producto']}) por ${total_venta(mejor):,}")

reporte(ventas)
```

# Glosario

| Término | En una línea |
|---|---|
| **Función** | Bloque de código con nombre, reutilizable, que puede recibir parámetros y devolver un valor |
| **`def`** | Palabra reservada para **definir** una función (no la ejecuta) |
| **Llamar** | Ejecutar una función: `nombre(argumentos)` |
| **Parámetro / argumento** | Variable en la definición / valor en la llamada |
| **`return`** | Entrega un valor y **termina** la función. Sin `return` → `None` |
| **Argumento posicional** | Se asigna por orden |
| **Argumento keyword** | Se asigna por nombre: `nombre=valor` |
| **Valor por defecto** | Parámetro opcional: `def f(x, y=0)` |
| **`*args`** | Captura posicionales sobrantes en una **tupla** |
| **`**kwargs`** | Captura keyword sobrantes en un **diccionario** |
| **Función vs método** | `len(x)` vs `x.upper()`: el método pertenece al objeto |
| **Orden superior** | Función que recibe o devuelve otra función |
| **`map`** | Aplica una función a cada elemento (transforma) |
| **`filter`** | Se queda con los elementos donde la función da `True` (selecciona) |
| **`lambda`** | Función anónima de **una expresión**: `lambda x: x * 2` |
| **`sorted` / `.sort()`** | Ordena: devuelve lista nueva / modifica la original y devuelve `None` |
| **`key=`** | Función que define el **criterio** de orden (o de máximo/mínimo) |
| **Módulo** | Archivo `.py` reutilizable |
| **Paquete** | Carpeta de módulos |
| **`import`, `as`, `from`** | Traer un módulo, darle un alias, traer solo una parte |
| **`pip` / `conda`** | Gestores de paquetes |
