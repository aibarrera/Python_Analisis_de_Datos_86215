## 2 · Variables y objetos
En Python **todo es un objeto**. Una **variable no es una caja**: es una **etiqueta (cartela) que apunta a un objeto**.

🏛️ Si dos cartelas apuntan a la misma obra, hay **una** obra con **dos** etiquetas.

- Una variable es una etiqueta que apunta a un objeto. Asignar es pegar una etiqueta, no copiar el contenido.

- **Confusión típica:** Pensar que `b = a` copia los datos. Copia la **referencia**. La documentación oficial lo dice así: la asignación simple no copia; un slice completo (`a[:]`) crea una copia superficial.

- **Documentación oficial:** [Modelo de datos: objetos, valores y tipos](https://docs.python.org/es/3/reference/datamodel.html#objects-values-and-types) · [Tutorial: listas, asignación y copias](https://docs.python.org/es/3/tutorial/introduction.html#lists)

```python
# TIPADO DINÁMICO: la variable es solo una etiqueta, así que puede apuntar a otro objeto más tarde.
a = 5                       # "a" apunta a un int
print(a)
a = "ahora soy texto"       # misma etiqueta, ahora apunta a un str. El 5 queda sin etiqueta.
print(a)
```

La misma etiqueta puede apuntar luego a otro tipo de dato (**tipado dinámico**).

### Dos etiquetas, un objeto

```python
# LISTA: colección ordenada y MUTABLE (se puede modificar). Se escribe entre corchetes.
obras_a = ["Paisaje", "Retrato"]
# Esta línea NO copia la lista: pega una SEGUNDA etiqueta al MISMO objeto (dos cartelas, una obra).
obras_b = obras_a
# append() agrega un elemento AL FINAL de la lista. Modifica el objeto original (mutabilidad).
obras_b.append("Taller")
# Como ambas etiquetas apuntan al mismo objeto, obras_a también muestra "Taller".
print(obras_a)
```

```python
# "is" compara IDENTIDAD: ¿son exactamente el mismo objeto en memoria? (no si tienen igual contenido)
print(obras_a is obras_b)             # True: hay un solo objeto con dos etiquetas
# id() devuelve el número de identidad del objeto; si coincide, es el mismo objeto.
print(id(obras_a), id(obras_b))
```

## 3 · Colecciones: guardar muchos datos juntos
Una **colección** agrupa varios datos en un solo objeto. Para elegir cuál usar nos preguntamos tres cosas: ¿tiene **orden**? ¿se puede **modificar**? ¿admite **repetidos**?

| Colección | Se escribe | Orden | Mutable | Repetidos | Para qué |
|---|---|---|---|---|---|
| **Lista** `list` | `[1, 2, 3]` | sí (por posición) | sí | sí | secuencia de elementos |
| **Tupla** `tuple` | `(1, 2)` | sí | **no** | sí | datos fijos |
| **Diccionario** `dict` | `{"clave": valor}` | por inserción | sí | claves únicas | **fichas**: campo → valor |
| **Conjunto** `set` | `{1, 2}` | **no** | sí | **no** | valores únicos |

🏛️ Lista = inventario. Tupla = ficha sellada. Diccionario = ficha de catálogo. Conjunto = lista de técnicas *distintas*.

- **Documentación oficial:** [Modelo de datos: valor y mutabilidad](https://docs.python.org/es/3/reference/datamodel.html#objects-values-and-types)

### 3.1 Listas

Una lista es una secuencia ordenada y mutable de elementos, que pueden repetirse y ser de distintos tipos. Se escribe entre corchetes. Se accede por posición (índice), que empieza en 0; los índices negativos cuentan desde el final.

Una lista es una colección ordenada que puedo modificar: agregar, quitar, cambiar.

- **Documentación oficial:** [Tutorial: más sobre listas](https://docs.python.org/es/3/tutorial/datastructures.html#more-on-lists)

```python
# Lista de cuatro elementos (str). Cada uno ocupa una posición (índice) que empieza en 0.
obras = ["Paisaje", "Retrato", "Taller", "Mercado"]
print(obras)
# len() devuelve la cantidad de elementos de una colección.
print(len(obras))
```

```python
# INDEXAR: obras[0] es el PRIMER elemento (los índices empiezan en 0).
print(obras[0])
# Índice negativo: cuenta desde el final. -1 es el último elemento.
print(obras[-1])
# SLICING [inicio:fin]: incluye el índice 1 y EXCLUYE el 3. Devuelve una lista nueva.
print(obras[1:3])
```

`obras[1:3]` es un **slicing**: incluye la posición 1 y **excluye** la 3.

```python
# Las listas son MUTABLES, por eso tienen métodos para agregar:
# append(x) agrega x AL FINAL de la lista.
obras.append("Autorretrato")
# insert(i, x) agrega x EN LA POSICIÓN i y corre los demás elementos hacia adelante.
obras.insert(1, "Río y luna")
print(obras)
```

```python
# remove(x) quita la PRIMERA aparición del valor x. Da ValueError si x no está en la lista.
obras.remove("Taller")
# pop() quita el ÚLTIMO elemento (o el de la posición que indiques) y lo DEVUELVE.
ultima = obras.pop()
print(ultima)
print(obras)
```

```python
# Modificar por índice: reemplaza el elemento de esa posición (posible porque la lista es mutable).
obras[0] = "Paisaje del sur"
# "in" pregunta por PERTENENCIA: devuelve True/False.
print("Mercado" in obras)
# count(x) cuenta cuántas veces aparece x. index(x) dice en qué posición está su primera aparición.
print(obras.count("Retrato"), obras.index("Retrato"))
```

```python
# sort() ordena la lista EN EL LUGAR (modifica la original) y devuelve None.
# Para textos, ordena alfabéticamente. Solo funciona si todos los elementos son comparables entre sí.
obras.sort()
print(obras)
```

### 3.2 Tuplas

Una tupla es una secuencia **ordenada** e **inmutable**. Se escribe entre paréntesis. Una tupla de un solo elemento necesita coma final: `(5,)`. Se puede *desempaquetar* en variables: `x, y = t`.

Una tupla es como una lista que no se puede modificar. Sirve para datos que no deberían cambiar.

- **Documentación oficial:** [Tutorial: tuplas y secuencias](https://docs.python.org/es/3/tutorial/datastructures.html#tuples-and-sequences)

```python
# TUPLA: colección ordenada e INMUTABLE. Se escribe entre paréntesis.
ubicacion = (2, 5)          # por ejemplo (sala, pared)
print(ubicacion[0])         # se accede por índice, igual que en una lista
# DESEMPAQUETADO: reparte los elementos en variables, en orden. La cantidad debe coincidir.
sala, pared = ubicacion
print(sala, pared)
```

```python
# ⚠️ Esta celda da error a propósito
# Una tupla es INMUTABLE: no admite asignar a un índice -> TypeError.
ubicacion[0] = 9
```

### 3.3 Diccionarios
Un diccionario guarda pares **clave → valor**. Es literalmente una **ficha de catálogo**.

Un diccionario asocia claves con valores. Las claves deben ser inmutables (cadenas, números, tuplas de inmutables) y son únicas. Se accede por clave, no por posición. Desde Python 3.7 conserva el orden de inserción. Pedir una clave que no existe con d[clave] da KeyError; d.get(clave) devuelve None (o el valor por defecto que indiques).

Un diccionario es una colección de pares: etiqueta y valor. Busco por etiqueta, no por posición.

- **Documentación oficial:** [Tutorial: diccionarios](https://docs.python.org/es/3/tutorial/datastructures.html#dictionaries)


```python
# DICCIONARIO: pares clave -> valor, entre llaves. Es una ficha de catálogo (campo: dato).
ficha = {"titulo": "Taller", "anio": 1985, "tecnica": "fotografía"}
print(ficha)
# Se accede por CLAVE (no por posición). Si la clave no existe, da KeyError.
print(ficha["titulo"])
```

```python
# Asignar a una clave NUEVA la agrega al diccionario.
ficha["expuesta"] = False
# Asignar a una clave EXISTENTE reemplaza el valor anterior (el 1985 se pierde).
ficha["anio"] = 1986
print(ficha)
```

```python
# get(clave) devuelve el valor, o None si la clave no existe (no da error como ficha["autor"]).
print(ficha.get("autor"))
# get(clave, valor_por_defecto) devuelve ese valor alternativo si la clave no existe.
print(ficha.get("autor", "sin dato"))
# "in" sobre un diccionario pregunta por las CLAVES.
print("anio" in ficha)
```

```python
# keys() devuelve las claves, values() los valores e items() los pares (clave, valor).
print(ficha.keys())
print(ficha.values())
print(ficha.items())
```

```python
# for recorre los pares que devuelve items(); en cada vuelta, "campo" toma la clave y "valor" el dato.
# La línea indentada (sangría de 4 espacios) es el bloque que se repite.
for campo, valor in ficha.items():
    print(campo, "->", valor)
```

### 3.4 Conjuntos

- **Definición, con palabras justas:** Un conjunto es una colección **sin orden** y **sin elementos repetidos**. Sirve para comprobar pertenencia y eliminar duplicados. Admite operaciones de conjuntos: unión `|`, intersección `&`, diferencia `-` y diferencia simétrica `^`.

Un conjunto guarda valores únicos. Si agrego uno que ya está, no pasa nada.

- **Documentación oficial:** [Tutorial: conjuntos](https://docs.python.org/es/3/tutorial/datastructures.html#sets)

```python
tecnicas = ["óleo", "grabado", "óleo", "acuarela", "grabado"]    # lista con repetidos
# SET (conjunto): colección SIN orden y SIN repetidos. set(lista) elimina los duplicados.
unicas = set(tecnicas)
print(unicas)             # el orden puede variar entre ejecuciones: los sets no tienen orden
print(len(unicas))        # 3 técnicas distintas
```

```python
# add(x) agrega un elemento al conjunto (si ya estaba, no pasa nada).
unicas.add("fotografía")
# discard(x) quita x si está. (remove(x) hace lo mismo, pero da error si x no existe.)
unicas.discard("grabado")
# "in" es muy eficiente en conjuntos: ideal para preguntar pertenencia.
print("óleo" in unicas)
print(unicas)
```

```python
sala_1 = {"óleo", "acuarela"}
sala_2 = {"acuarela", "grabado"}
# & INTERSECCIÓN: lo que está en ambos conjuntos.
print(sala_1 & sala_2)
# | UNIÓN: todo lo que está en alguno de los dos (sin repetir).
print(sala_1 | sala_2)
# - DIFERENCIA: lo que está en sala_1 pero NO en sala_2.
print(sala_1 - sala_2)
```

## 4 · Operaciones y control de flujo (primer vistazo)
Los operadores combinan datos; las estructuras de control deciden **qué se ejecuta y cuántas veces**. Hoy vemos lo esencial; en la próxima clase profundizamos.

** `+ - *` operan como en matemática; `/` siempre devuelve `float`; `//` es división entera; `%` es el resto; `**` es potencia. Tienen orden de precedencia: `**` primero, luego `* / // %`, luego `+ -`; los paréntesis lo alteran. Los operadores de comparación (`== != < <= > >=`) devuelven siempre un booleano; `in` comprueba pertenencia.

Hago cuentas con los operadores aritméticos y hago preguntas de sí o no con los de comparación.

- **Documentación oficial:** [Tutorial: Python como calculadora](https://docs.python.org/es/3/tutorial/introduction.html#using-python-as-a-calculator)

```python
print(7 + 3, 7 - 3, 7 * 3)                  # suma, resta, producto: 10 4 21
# "/" siempre da float (3.5). "//" es división ENTERA (3). "%" es el RESTO (1). "**" es potencia (1024).
print(7 / 2, 7 // 2, 7 % 2, 2 ** 10)
# PRECEDENCIA: primero ** , luego * / // % , luego + - . Los paréntesis cambian el orden.
print(2 + 3 * 4, (2 + 3) * 4)               # 14 y 20
```

```python
# Comparaciones: devuelven siempre un booleano (True/False). "==" compara valores; "=" asigna.
print(5 == 5, 5 != 5, 5 > 3)                # igual, distinto, mayor
print("Python" in ["Python", "SQL"])         # "in": pertenencia a una colección
anios = 4
# and: ambas condiciones deben ser True. (or: alcanza con una. not: invierte el resultado.)
print(anios >= 2 and anios < 5)
```

`and`, `or` y `not` combinan condiciones. Con textos y listas, `+` une y `*` repite:

```python
# En textos y listas, + UNE (concatena) y * REPITE.
print("Data" + " " + "Science")     # une tres textos
print("-" * 10)                      # repite el texto 10 veces
print([1, 2] + [3])                  # une dos listas en una nueva: [1, 2, 3]
```

### Condicional: `if` / `elif` / `else`
La **sangría** (4 espacios) marca qué pertenece al bloque. 🏛️ Es el criterio curatorial.

El condicional ejecuta un bloque solo si se cumple una condición. Puede haber cero o más `elif` y un `else` opcional. Python no tiene `switch`: una cadena `if`-`elif`-`else` lo reemplaza. Los bloques se delimitan por **indentación**.

Si pasa esto, hago aquello; si no, hago otra cosa. Python usa la sangría para saber qué está adentro.

- **Documentación oficial:** [Tutorial: la sentencia if](https://docs.python.org/es/3/tutorial/controlflow.html#if-statements)

```python
anio = 1952
# if / elif / else: se evalúan en orden y se ejecuta SOLO el primer bloque cuya condición sea True.
# Cada condición termina en ":" y su bloque va con sangría de 4 espacios.
if anio < 1920:
    print("Sala histórica")
elif anio < 1970:                  # solo se evalúa si la anterior fue False
    print("Sala de mitad de siglo")
else:                              # se ejecuta si ninguna condición anterior se cumplió
    print("Sala contemporánea")
```

Para cada elemento de la colección, hago esto.

- **Documentación oficial:** [Tutorial: la sentencia for](https://docs.python.org/es/3/tutorial/controlflow.html#for-statements)


```python
# for recorre la lista; en cada vuelta, "obra" apunta a una ficha (un diccionario) distinta.
# Se ejecuta tantas veces como elementos tenga "coleccion". Es como inventariar obra por obra.
for obra in coleccion:
    print(obra["titulo"], obra["anio"])
```

### `while`, `break` y `continue`

```python
ahorro = 0
meses = 0
while ahorro < 1000:
    ahorro += 150
    meses += 1
print(meses)
```

```python
for n in range(1, 11):
    if n % 2 == 0:
        continue
    if n > 7:
        break
    print(n)
```

## 📚 Glosario rápido
| Término | En una línea |
|---|---|
| **Dato** | Representación simbólica de un aspecto de la realidad |
| **Tipo** | Define qué valores y operaciones admite un dato (`int`, `float`, `str`, `bool`) |
| **Variable** | Nombre (etiqueta) que apunta a un objeto |
| **Objeto** | Dato con identidad, tipo y valor |
| **Mutable / inmutable** | Que se puede / no se puede modificar sin crear otro objeto |
| **Colección** | Objeto que agrupa varios datos (`list`, `tuple`, `dict`, `set`) |
| **Índice** | Posición de un elemento (empieza en 0) |
| **Clave** | Etiqueta con la que se busca un valor en un diccionario |
| **Bloque** | Código con sangría que pertenece a un `if` o a un `for` |
| **Notebook** | Documento que mezcla texto, código y resultados |
